RiskTwin veri kaynaklarının yüklenmesi ve temel istatistiklerin incelenmesi.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt

from src.data_ingestion import load_hazard_data, load_building_data, load_population_data
from src.data_processing import validate_hazard_data, validate_building_data, clean_building_data
```

## 1. Tehlike Verisi (AFAD)

```python
hazard_df = load_hazard_data()
print(f'Kayıt sayısı: {len(hazard_df)}')
print(f'Eksik kolonlar: {validate_hazard_data(hazard_df)}')
hazard_df.head()
```

```python
hazard_df.describe()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.barh(hazard_df['location_name'], hazard_df['hazard_score'], color='#e74c3c')
ax.set_xlabel('Tehlike Skoru')
ax.set_title('Lokasyon Bazlı AFAD Tehlike Skorları')
plt.tight_layout()
plt.show()
```

## 2. Bina Verisi

```python
building_df = load_building_data()
building_df = clean_building_data(building_df)
print(f'Kayıt sayısı: {len(building_df)}')
print(f'Eksik kolonlar: {validate_building_data(building_df)}')
building_df.head()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

building_df['structural_system'].value_counts().plot(kind='bar', ax=axes[0], color='#3498db')
axes[0].set_title('Taşıyıcı Sistem Dağılımı')

building_df['soil_class'].value_counts().plot(kind='bar', ax=axes[1], color='#f39c12')
axes[1].set_title('Zemin Sınıfı Dağılımı')

building_df['building_age'].hist(ax=axes[2], bins=10, color='#27ae60')
axes[2].set_title('Bina Yaşı Dağılımı')

plt.tight_layout()
plt.show()
```
