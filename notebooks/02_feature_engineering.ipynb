```python
import sys
sys.path.insert(0, '..')

import pandas as pd
from src.data_ingestion import load_hazard_data, load_building_data
from src.data_processing import clean_building_data, merge_hazard_building
from src.feature_engineering import encode_features, prepare_training_data, get_feature_names
```

## 1. Veri Hazırlama

```python
hazard_df = load_hazard_data()
building_df = load_building_data()
building_df = clean_building_data(building_df)
merged_df = merge_hazard_building(hazard_df, building_df)
print(f'Birleştirilmiş veri: {merged_df.shape}')
merged_df.head()
```

## 2. Feature Encoding

```python
features = encode_features(merged_df)
print(f'Feature matrix boyutu: {features.shape}')
print(f'Feature isimleri: {list(features.columns)}')
features.head()
```

## 3. Training Data Hazırlama

```python
X, y = prepare_training_data(merged_df)
print(f'X shape: {X.shape}')
print(f'y dağılımı:\n{y.value_counts()}')
print(f'\nTemel feature isimleri: {get_feature_names()}')
```

```python
# Feature korelasyon matrisi
base_features = get_feature_names()
available = [f for f in base_features if f in X.columns]
X[available].corr().style.background_gradient(cmap='RdYlGn_r')
```
